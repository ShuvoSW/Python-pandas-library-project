```python
import pandas as pd
import numpy as np
```

```python
data = pd.DataFrame({
    'Name': ['Alice', 'Bob', np.nan, 'Daisy', 'Ethan'],
    'Age': [25, np.nan, 29,22,np.nan],
    'Salary': [50000, 60000, np.nan, 58000, 62000]
})
```

```python
data
```

```python
data_dropped = data.dropna()
```

```python
data_dropped
```

```python
data_filled = data.fillna({
    'Name': 'Unknown',
    'Age': data['Age'].mean(),
    'Salary': 0
})
```

```python
data_filled
```
