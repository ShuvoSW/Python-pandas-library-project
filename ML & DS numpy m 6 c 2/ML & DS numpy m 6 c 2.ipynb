```python
# vertical stack
# horizontal stack
# depth stack
```

```python
a = np.array([[1,2], [3,4]])
b = np.array([[5,6]])
```

```python
v_stacked = np.vstack([a, b])
print(v_stacked)
```

```python
a = np.array([[1,2,9], [3,4,0]])
b = np.array([[5,6], [7, 8]])
```

```python
h_stacked = np.hstack([a, b])
print(h_stacked)
```

```python
print(a)
print(b)
```

```python
a = np.array([[1,2,], [3,4,]])
b = np.array([[5,6], [7, 8]])
```

```python
d_stacked = np.dstack([a, b])
print(d_stacked)
```

```python
a = np.array([[1,2,]])
b = np.array([[5,6]])
```

```python
d_stacked = np.dstack([a, b])
print(d_stacked)
```
